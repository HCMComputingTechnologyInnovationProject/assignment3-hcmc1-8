```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")
```

```python
ulb = pd.read_csv("../data/processed/processed_ulb_dataset.csv")
ieee = pd.read_csv("../data/processed/processed_ieee_dataset.csv")
paysim = pd.read_csv("../data/processed/processed_paysim_dataset.csv")

print("ULB dataset shape:", ulb.shape)
print("IEEE dataset shape:", ieee.shape)
print("PaySim dataset shape:", paysim.shape)
```

```python
print("ULB Dataset Info")
ulb.info()

print("\nIEEE Dataset Info")
ieee.info()

print("\nPaySim Dataset Info")
paysim.info()
```

```python
datasets = {
    "ULB": ulb,
    "IEEE": ieee,
    "PaySim": paysim
}

for name, df in datasets.items():
    
    plt.figure(figsize=(5,4))
    
    sns.countplot(x=df["Class"] if "Class" in df.columns else df["isFraud"])
    
    plt.title(f"{name} Fraud Distribution")
    
    plt.show()
```

```python
for name, df in datasets.items():
    
    target = "Class" if "Class" in df.columns else "isFraud"
    
    fraud_ratio = df[target].value_counts(normalize=True)
    
    print(name)
    print(fraud_ratio)
    print()
```
